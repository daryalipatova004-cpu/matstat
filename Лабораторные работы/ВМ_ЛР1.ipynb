2. Реализовать построение выборочной функции распределения для генерируемой автоматически выборки размером $n$ = $10^{1}$, $10^{2}$, $10^{3}$, $10^{4}$ значений.

```python
import matplotlib.pyplot as plt
from scipy.stats import ecdf

samples = [np.random.rand(i) for i in [10**1, 10**2, 10**3, 10**4]]
fig, axs = plt.subplots(4, figsize=(10, 18))
fig.suptitle('Выборочные функции распределения')
for i in range(len(samples)):
    res = ecdf(samples[i])
    res.cdf.plot(axs[i])
    axs[i].set_xlabel('X')
    axs[i].set_ylabel('ECDF')
    axs[i].set_title(f'Размер выборки: 10**{i+1}')
plt.tight_layout(rect=[0, 0, 1, 0.98])
plt.show()
```

3. Реализовать построение гистограммы для генерируемой автоматически выборки
размером $n$ = $10^{1}$, $10^{2}$, $10^{3}$, $10^{4}$ значений. Количество полос определять по соотношению $n/10$,
ширину полос выбирать одинаковой.

```python
fig, axs = plt.subplots(4, figsize=(10, 18))
fig.suptitle('Гистограммы')
for i in range(len(samples)):
    axs[i].hist(samples[i], len(samples[i]) // 10)
    axs[i].set_xlabel('X')
    axs[i].set_ylabel('Количество')
    axs[i].set_title(f'Размер выборки: 10**{i+1}')
plt.tight_layout(rect=[0, 0, 1, 0.98])
plt.show()
```
