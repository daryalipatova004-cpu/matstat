Нужно составить максимально лаконичный код, который осуществлял бы оценку интересующего параметра по заданной выборки для плотности распределения, заданной отдельной функцией. Чем короче код, тем лучше.


Спецификация программы предполагается следующей: есть процедура, реализующая метод максимального правдоподобия в численной форме (оптимизация), условно назовем ее p_est = MLE(@pdf, x, p0).

Ей на вход передается указатель или ссылка @pdf на функцию pdf, которая реализует вычисление значения плотности распределения: y = pdf(t, p), где t - аргумент плотности, p - вектор параметров, от которых зависит плотность и для которых производится вычисление значения оценки методом максимального правдоподобия, y - возвращаемое значение плотности.

Вектор p0 - это начальные приближения оценок параметров (например, все нули - это значения, с которых начинается поиск оценок при численной оптимизации).

Вектор p_est - это возвращенные значения оценок.

```python
import numpy as np
from scipy.optimize import minimize

# Например, функция для нормального распределения
def pdf(x, p):
    mu, s = p
    return np.exp(-(x-mu)**2/(2*s**2))/(s*np.sqrt(2*np.pi))

def MLE(pdf, x, p0):
    return minimize(lambda p: -np.log(pdf(x, p)).sum(), p0).x

x = np.array([1.2, 0.8, 1.5, 1.1, 0.9])
p_est = MLE(pdf, x, [0.0, 1.0])
print(p_est)
```
