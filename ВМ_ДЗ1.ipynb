Необходимо составить максимально лаконичный код, который осуществлял бы построение выборочной функции распределения для заданной выборки и производил бы оценку выборочной квантили для заданной вероятности. Чем короче код, тем лучше

Тут коротко в ущерб читаемости графика:

```python
import matplotlib.pyplot as plt
from scipy.stats import ecdf, quantile

sample = [1, 1, 4, 3, 2, 8, 3, 9, 0, 4, 6, 5]
p = 0.5

print(quantile(sample, p))
ecdf(sample).cdf.plot(plt.subplot())
plt.show()
```

Тут чуть длиннее и чуть более читаемо:

```python
import matplotlib.pyplot as plt
from scipy.stats import ecdf, quantile

sample = [1, 1, 4, 3, 2, 8, 3, 9, 0, 4, 6, 5]
p = 0.5

print(f'{p * 100}% квантиль: {quantile(sample, p)}')

res = ecdf(sample)
ax = plt.subplot()
res.cdf.plot(ax)
ax.set_xlabel('X')
ax.set_ylabel('ECDF')
ax.set_title('Выборочная функция распределения')
plt.show()
```
